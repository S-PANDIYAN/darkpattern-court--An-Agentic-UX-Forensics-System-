# Dark Pattern Detection - Baseline Report
**Deliverable for Week 3**

This notebook covers the exploratory data analysis, class imbalance visualisations, and the initial TF-IDF baseline results for our dark pattern detection project.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import json

# Load the dataset
df = pd.read_csv('lumichats_master_dataset_v2.csv', keep_default_na=False)
print(f"Total samples: {len(df)}")
df.head()
```

## Class Distribution and Imbalance

```python
plt.figure(figsize=(10, 6))
sns.countplot(y='label', data=df, order=df['label'].value_counts().index)
plt.title('Dark Pattern Category Distribution')
plt.tight_layout()
plt.show()
```

## TF-IDF Baseline Results
We trained a Logistic Regression model on TF-IDF features as our baseline. Let's load and review the results.

```python
with open('baseline_results.json', 'r') as f:
    baseline = json.load(f)

print("Baseline Metrics on Test Set:")
print(f"Macro F1: {baseline['macro_f1']:.4f}")
print(f"Weighted F1: {baseline['weighted_f1']:.4f}")
print(f"Accuracy: {baseline['accuracy']:.4f}")

print("\nPer-Class F1 Scores:")
for cls, score in baseline['per_class_f1'].items():
    print(f"- {cls}: {score:.4f}")
```
